## Part 1. The basics
### Your first line of Python
`print()` shows something on the screen. Text goes inside quotes. This is the classic first program.

```python
print("Hello, world!")
print("Welcome to the Python crash course.")
```

`print()` can show several things at once, separated by commas. Python puts a space between them.
You can also do maths right inside it.

```python
print("Cups of coffee today:", 3)
print("Minutes in a day:", 24 * 60)
```

### Variables
A **variable** is a labelled box that stores a value. You create one with `=`.
Here we describe a pet. Later, we can use the name instead of retyping the value.

```python
pet_name = "Biscuit"
pet_age = 4
pet_weight_kg = 7.5
is_vaccinated = True

print(pet_name, "is", pet_age, "years old")
```

### Types
Every value has a **type**. The four basic ones are:
- `str`: text, like `"Biscuit"`
- `int`: whole numbers, like `4`
- `float`: decimal numbers, like `7.5`
- `bool`: `True` or `False`

`type()` tells you which one a value is.

```python
print(type(pet_name))
print(type(pet_age))
print(type(pet_weight_kg))
print(type(is_vaccinated))
```

Variables can change. Here a phone battery goes down as we use it.
The line `battery = battery - 15` means: take the old value, subtract 15, and store the result back.

```python
battery = 100
print("Morning:", battery)
battery = battery - 15
print("After music:", battery)
battery -= 20          # a shortcut for battery = battery - 20
print("After video calls:", battery)
```

### Strings
Text is called a **string**. Strings have built-in **methods**: little tools you call with a dot.
Here we tidy up a messy song title.

```python
song = "  bohemian rhapsody  "
print(song.strip())            # remove spaces at both ends
print(song.strip().title())    # Capitalise Each Word
print(song.upper())            # ALL CAPITALS
print(len(song.strip()), "characters")
```

You can **join** strings with `+`, and **pick** characters by position. Positions start at **0**, not 1.
A negative position counts from the end.

```python
first = "Jollof"
second = "rice"
dish = first + " " + second
print(dish)
print(dish[0])      # first character
print(dish[-1])     # last character
print(dish[0:6])    # a slice: positions 0 up to (not including) 6
```

### f-strings
An **f-string** is the easiest way to put values inside text. Put `f` before the quotes, and variables inside `{ }`.
You can even format numbers, for example `:.2f` means two decimal places.

```python
item = "latte"
price = 3.5
cups = 2
print(f"{cups} x {item} costs ${price * cups:.2f}")
print(f"{pet_name} weighs {pet_weight_kg} kg")
```

### Numbers and maths
Python is a great calculator. The main operators are `+ - * /`, plus three special ones:
- `//` divides and drops the remainder
- `%` gives only the remainder
- `**` is "to the power of"

Here we split 17 slices of pizza between 5 friends.

```python
slices = 17
friends = 5
print("Each friend gets:", slices // friends)
print("Left over:", slices % friends)
print("Exact share:", slices / friends)
print("2 to the power of 10:", 2 ** 10)
```

`round()` tidies decimals, and `int()`, `float()` and `str()` **convert** between types.
This matters because `"5" + "5"` joins text, while `5 + 5` adds numbers.

```python
temperature = 23.678
print(round(temperature, 1))
print("5" + "5")
print(int("5") + int("5"))
print(str(42) + " is the answer")
```

## Part 2. Collections
### Lists
A **list** holds many values in order, inside square brackets. Lists can grow and shrink.
Here is a shopping list.

```python
shopping = ["milk", "eggs", "bread"]
print(shopping)
print("First item:", shopping[0])
print("Number of items:", len(shopping))
```

Lists have handy methods: `append` adds to the end, `remove` takes an item out, `sort` puts them in order.
`in` checks whether something is in the list.

```python
shopping.append("apples")
shopping.append("cheese")
shopping.remove("eggs")
shopping.sort()
print(shopping)
print("Do we need milk?", "milk" in shopping)
print("Do we need eggs?", "eggs" in shopping)
```

### Tuples
A **tuple** is like a list, but it can't be changed after you create it. It uses round brackets.
It's perfect for things that belong together and stay fixed, like a map location.
You can **unpack** a tuple into separate variables in one line.

```python
lagos = (6.52, 3.38)          # (latitude, longitude)
latitude, longitude = lagos
print(f"Latitude {latitude}, longitude {longitude}")

try:
    lagos[0] = 0
except TypeError as e:
    print("Can't change a tuple:", e)
```

### Dictionaries
A **dictionary** stores **key: value** pairs, like a real dictionary stores word: meaning.
You look things up by key instead of by position. Here is a small café menu.

```python
menu = {"espresso": 2.5, "latte": 3.5, "tea": 2.0}
print(menu["latte"])
menu["mocha"] = 4.0        # add a new item
menu["tea"] = 2.2          # change a price
print(menu)
print(menu.get("smoothie", "not on the menu"))
```

You can loop over a dictionary with `.items()`, which gives you each key and value together.

```python
for drink, cost in menu.items():
    print(f"{drink:<10} ${cost:.2f}")
```

### Sets
A **set** keeps only **unique** values, with no duplicates and no order.
Sets are great for questions like "who came to both parties?".

```python
friday = {"Ada", "Ben", "Chi", "Dayo"}
saturday = {"Chi", "Dayo", "Ema"}
print("Both nights:", friday & saturday)
print("Anyone at all:", friday | saturday)
print("Only Friday:", friday - saturday)
print(set(["red", "blue", "red", "red"]))
```

## Part 3. Control flow
### if / elif / else
`if` lets your program make decisions. Python checks each condition from top to bottom and runs the **first** one that's true.
**Indentation** (the spaces at the start of a line) shows which lines belong to each branch.

```python
temperature = 31

if temperature >= 30:
    print("Hot! Drink water and wear a hat.")
elif temperature >= 18:
    print("Nice weather for a walk.")
else:
    print("Bring a jacket.")
```

Conditions can be combined with `and`, `or` and `not`. Here: can we go to the beach?

```python
is_weekend = True
is_raining = False
money = 15

if is_weekend and not is_raining and money >= 10:
    print("Beach day!")
else:
    print("Maybe next time.")
```

### for loops
A `for` loop repeats code for **each item** in a collection. Here we print a playlist with track numbers.
`enumerate` gives us a counter alongside each item.

```python
playlist = ["Calm Down", "Essence", "Water", "Love Nwantiti"]
for number, track in enumerate(playlist, start=1):
    print(f"{number}. {track}")
```

`range()` makes a sequence of numbers, which is handy for counting. Here we count steps on a daily walk goal.

```python
total_steps = 0
for day in range(1, 6):
    steps = day * 2000
    total_steps += steps
    print(f"Day {day}: {steps} steps")
print("Total:", total_steps)
```

### while loops
A `while` loop repeats **as long as** a condition is true. Here we keep saving money until we reach a goal.
`break` stops a loop early, and `continue` skips to the next round.

```python
savings = 0
week = 0
while savings < 100:
    week += 1
    savings += 30
    print(f"Week {week}: saved ${savings}")
print("Goal reached in", week, "weeks")
```

Here `continue` skips the rest of a round, and `break` stops the loop completely.

```python
for number in range(1, 20):
    if number % 2 == 0:
        continue           # skip even numbers
    if number > 9:
        break              # stop after 9
    print(number, end=" ")
```

### Comprehensions
A **list comprehension** builds a new list in one readable line: *"give me this, for each item, if a condition holds"*.
Here we convert temperatures and pick out the warm days.

```python
celsius = [12, 19, 25, 31, 8]
fahrenheit = [c * 9 / 5 + 32 for c in celsius]
warm_days = [c for c in celsius if c >= 20]
print(fahrenheit)
print(warm_days)
```

Comprehensions work for dictionaries too. Here we make a price list with a 10% discount.

```python
prices = {"rice": 20, "beans": 12, "oil": 15}
sale = {item: round(cost * 0.9, 2) for item, cost in prices.items()}
print(sale)
```

## Part 4. Functions
### Defining a function
A **function** is a reusable recipe. You write it once with `def`, and call it as often as you like.
`return` sends a result back to whoever called it.

```python
def tip(bill, percent):
    return bill * percent / 100

print(tip(40, 10))
print(tip(85, 15))
```

### Default values
Give a parameter a **default**, and callers can skip it. Here the default greeting is "Hello".
You can also pass arguments **by name**, which makes calls easier to read.

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))
print(greet("Ben", "Good morning"))
print(greet(greeting="Welcome back", name="Chi"))
```

### Returning several values
A function can return several values at once. Python packs them into a tuple, and you can unpack them.

```python
def min_max_average(scores):
    return min(scores), max(scores), sum(scores) / len(scores)

low, high, avg = min_max_average([72, 88, 95, 61])
print(f"Lowest {low}, highest {high}, average {avg}")
```

### *args and **kwargs
`*args` collects **any number** of positional arguments into a tuple.
`**kwargs` collects **any number** of named arguments into a dictionary.
Here: a pizza order with any toppings, and any extra options.

```python
def order_pizza(size, *toppings, **options):
    print(f"A {size} pizza with {', '.join(toppings) or 'no toppings'}")
    for key, value in options.items():
        print(f"  {key}: {value}")

order_pizza("large", "mushrooms", "peppers", "olives", crust="thin", delivery=True)
order_pizza("small")
```

### Scope
Variables created **inside** a function only exist inside it. That's called **local scope**.
Variables outside are **global**. A function can read them, but it's clearer to pass values in and return results.

```python
tax_rate = 0.075          # global

def price_with_tax(price):
    tax = price * tax_rate   # local: only exists inside the function
    return price + tax

print(price_with_tax(200))
try:
    print(tax)
except NameError as e:
    print("Outside the function:", e)
```

## Part 5. Errors
### try / except
When something goes wrong, Python raises an **error** (an exception) and stops.
`try / except` lets you **catch** the error and handle it calmly. Here we split a restaurant bill.

```python
def split_bill(total, people):
    try:
        return total / people
    except ZeroDivisionError:
        print("You can't split a bill between 0 people!")
        return None

print(split_bill(90, 3))
print(split_bill(90, 0))
```

You can catch different errors separately. `else` runs when nothing went wrong, and `finally` **always** runs.
Here we read a person's age from text, which might not be a number.

```python
for text in ["34", "thirty", "-5"]:
    try:
        age = int(text)
        if age < 0:
            raise ValueError("age can't be negative")
    except ValueError as e:
        print(f"{text!r}: invalid ({e})")
    else:
        print(f"{text!r}: valid age {age}")
    finally:
        print("  checked", text)
```

### Raising your own errors
`raise` lets **your** code signal a problem. Here a bank withdrawal refuses to go below zero.

```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError(f"not enough money: balance {balance}, asked for {amount}")
    return balance - amount

print(withdraw(100, 30))
try:
    withdraw(100, 250)
except ValueError as e:
    print("Refused:", e)
```

## Part 8. Classes and objects (OOP)
### Your first class
A **class** is a blueprint. An **object** is one thing built from it.
`__init__` runs when you create an object and stores its data on `self`. **Methods** are functions that belong to the class.
Here: a dog.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        self.tricks = []

    def learn(self, trick):
        self.tricks.append(trick)

    def show_off(self):
        return f"{self.name} the {self.breed} can: {', '.join(self.tricks)}"

rex = Dog("Rex", "beagle")
rex.learn("sit")
rex.learn("roll over")
print(rex.show_off())
```

Each object keeps its **own** data. Here a simple bank account protects its balance with a method.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("insufficient funds")
        self.balance -= amount
        return self.balance

mine = BankAccount("Ada", 50)
yours = BankAccount("Ben")
mine.deposit(25)
yours.deposit(10)
print(mine.owner, mine.balance, "|", yours.owner, yours.balance)
```

### Inheritance
A class can **inherit** from another, reusing its code and changing only what's different.
`super()` calls the parent's version. Here, a savings account is a bank account that also earns interest.

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=0.05):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.balance = round(self.balance * (1 + self.rate), 2)
        return self.balance

nest_egg = SavingsAccount("Chi", 1000)
nest_egg.deposit(200)          # inherited from BankAccount
print(nest_egg.add_interest())  # new in SavingsAccount
```

### Dunder methods
Methods with **double underscores** ("dunder") let your objects work with Python's built-in features.
`__str__` controls how it prints, `__len__` makes `len()` work, and `__add__` makes `+` work.
Here: a music playlist.

```python
class Playlist:
    def __init__(self, name, songs):
        self.name = name
        self.songs = list(songs)

    def __str__(self):
        return f"Playlist '{self.name}' ({len(self)} songs)"

    def __len__(self):
        return len(self.songs)

    def __add__(self, other):
        return Playlist(f"{self.name} + {other.name}", self.songs + other.songs)

gym = Playlist("Gym", ["Run", "Power"])
chill = Playlist("Chill", ["Sunset", "Waves", "Rain"])
print(gym)
print(gym + chill)
```

### Dataclasses
For classes that mainly **hold data**, `@dataclass` writes `__init__`, a nice printout, and comparisons for you.
Here: books on a shelf, sorted by year.

```python
from dataclasses import dataclass

@dataclass
class Book:
    title: str
    author: str
    year: int

shelf = [Book("Things Fall Apart", "Chinua Achebe", 1958),
         Book("Half of a Yellow Sun", "Chimamanda Ngozi Adichie", 2006),
         Book("The Famished Road", "Ben Okri", 1991)]
print(shelf[0])
print(Book("A", "B", 1) == Book("A", "B", 1))
```

## Part 9. Advanced tools
### Iterators
Anything you can loop over is **iterable**. `iter()` gets an iterator, and `next()` takes one item at a time.
This is what a `for` loop does behind the scenes. Here: a queue at a bakery.

```python
queue = iter(["Ada", "Ben", "Chi"])
print("Now serving:", next(queue))
print("Now serving:", next(queue))
print("Now serving:", next(queue))
try:
    next(queue)
except StopIteration:
    print("The queue is empty")
```

### Generators
A **generator** is a function that uses `yield` to hand out values **one at a time**, only when asked.
It doesn't build the whole list in memory. Here: a countdown for a rocket launch.

```python
def countdown(start):
    while start > 0:
        yield start
        start -= 1
    yield "Lift off!"

for value in countdown(5):
    print(value, end=" ")
```

Generators can even be endless, because they only make the next value when you need it.
Here we generate ticket numbers forever, but only take the first five.

```python
def ticket_numbers(prefix):
    n = 1
    while True:
        yield f"{prefix}-{n:03d}"
        n += 1

tickets = ticket_numbers("BUS")
print([next(tickets) for _ in range(5)])
squares = (n * n for n in range(1, 6))     # a generator expression
print(sum(squares))
```

### lambda, map, filter and sorted
A **lambda** is a tiny, nameless function written in one line.
`map` applies a function to every item, `filter` keeps items that pass a test,
and `sorted(..., key=...)` sorts by whatever you choose. Here: movies and their ratings.

```python
movies = [("Inception", 8.8), ("Cats", 2.8), ("Up", 8.3), ("Coco", 8.4)]

print(sorted(movies, key=lambda m: m[1], reverse=True))
good = list(filter(lambda m: m[1] >= 8, movies))
print(good)
print(list(map(lambda m: m[0].upper(), movies)))
```

`sorted` with `key` also works on dictionaries and strings. Here we sort friends by age, then names by length.

```python
friends = [{"name": "Ada", "age": 29}, {"name": "Ben", "age": 24}, {"name": "Chi", "age": 31}]
for f in sorted(friends, key=lambda f: f["age"]):
    print(f["name"], f["age"])
print(sorted(["kiwi", "banana", "fig", "apple"], key=len))
```

### Decorators
A **decorator** wraps a function to add extra behaviour, without changing the function itself.
You apply it with `@`. Here, a decorator that announces each kitchen task and times it.

```python
import time
from functools import wraps

def announce(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Starting {func.__name__}...")
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__} in {time.perf_counter() - start:.2f}s")
        return result
    return wrapper

@announce
def boil_eggs(n):
    time.sleep(0.2)
    return f"{n} eggs boiled"

print(boil_eggs(3))
```

Decorators can be reused on any function. Here the same idea counts how many times a doorbell was pressed.

```python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper

@count_calls
def ring_doorbell():
    return "Ding dong!"

for _ in range(3):
    print(ring_doorbell())
print("Pressed", ring_doorbell.calls, "times")
```

### Context managers
A **context manager** sets something up, and **always** cleans it up afterwards, even if there's an error.
You already used one: `with open(...)`. You can make your own with `@contextmanager`.
Here: lights that switch off when you leave a room.

```python
from contextlib import contextmanager

@contextmanager
def room(name):
    print(f"Lights ON in the {name}")
    try:
        yield name
    finally:
        print(f"Lights OFF in the {name}")

with room("kitchen") as r:
    print(f"Cooking in the {r}")

try:
    with room("garage"):
        raise RuntimeError("tripped over a box")
except RuntimeError as e:
    print("Problem:", e)
```

## Part 10. Type hints and testing
### Type hints
**Type hints** note what kinds of values a function expects and returns. Python doesn't enforce them,
but they make code easier to read, and editors use them to catch mistakes early.

```python
def bmi(weight_kg: float, height_m: float) -> float:
    return round(weight_kg / height_m ** 2, 1)

def shopping_total(prices: dict[str, float], discount: float = 0.0) -> float:
    return round(sum(prices.values()) * (1 - discount), 2)

print(bmi(70, 1.75))
print(shopping_total({"rice": 20.0, "fish": 35.5}, discount=0.1))
print(bmi.__annotations__)
```

Hints can say "this or nothing" with `| None`, and describe lists of things with `list[...]`.

```python
def find_contact(contacts: list[str], name: str) -> str | None:
    for c in contacts:
        if c.lower().startswith(name.lower()):
            return c
    return None

phone_book = ["Ada Obi", "Ben Cole", "Chi Eze"]
print(find_contact(phone_book, "ben"))
print(find_contact(phone_book, "zara"))
```

### Testing with assert
`assert` checks that something is true. If it isn't, Python raises an `AssertionError`.
Writing a few asserts is the simplest way to test your functions. Here we test the `tip` function from Part 4.

```python
assert tip(100, 10) == 10
assert tip(50, 20) == 10
assert greet("Ada") == "Hello, Ada!"
assert bmi(70, 1.75) == 22.9
print("All tests passed!")
```

When a test fails, you see exactly which check broke. Here we test a function with a **bug** on purpose.

```python
def average(numbers):
    return sum(numbers) / (len(numbers) + 1)      # bug: + 1 should not be there

try:
    assert average([2, 4, 6]) == 4, f"expected 4, got {average([2, 4, 6])}"
except AssertionError as e:
    print("Test failed:", e)
```

First, load the data. On Colab, the file is downloaded once from the course's GitHub repo.

```python
import csv, pathlib, urllib.request

DATA = pathlib.Path("data/expenses.csv")
if not DATA.exists():
    DATA.parent.mkdir(exist_ok=True)
    urllib.request.urlretrieve(
        "https://raw.githubusercontent.com/davolu/python-crash-course/main/data/expenses.csv", DATA)

print(DATA.read_text()[:300])
```

We describe one expense with a **dataclass**, and read every row of the CSV into a list of them.
Notice the type hints, and that we convert the amount from text to a `float`.

```python
from dataclasses import dataclass

@dataclass
class Expense:
    date: str
    category: str
    description: str
    amount: float

def load_expenses(path: pathlib.Path) -> list[Expense]:
    with open(path, newline="") as f:
        return [Expense(r["date"], r["category"], r["description"], float(r["amount"]))
                for r in csv.DictReader(f)]

expenses = load_expenses(DATA)
print(len(expenses), "expenses loaded")
print(expenses[0])
```

Now the totals. We add up each category in a **dictionary**, then sort the categories by total,
biggest first, with `sorted` and a `lambda`.

```python
def totals_by_category(items: list[Expense]) -> dict[str, float]:
    totals: dict[str, float] = {}
    for e in items:
        totals[e.category] = totals.get(e.category, 0) + e.amount
    return {k: round(v, 2) for k, v in sorted(totals.items(), key=lambda kv: kv[1], reverse=True)}

totals = totals_by_category(expenses)
for category, amount in totals.items():
    print(f"{category:<12} ${amount:>7.2f}")
print(f"{'TOTAL':<12} ${sum(totals.values()):>7.2f}")
```

Next, the three biggest single expenses, using `sorted` with a `key`, and a **slice** to take the top 3.

```python
top3 = sorted(expenses, key=lambda e: e.amount, reverse=True)[:3]
for e in top3:
    print(f"{e.date}  {e.description:<16} ${e.amount:.2f}")
```

A budget check: we set a budget per category, and use a comprehension to find where we went over.
A **generator** gives us a warning line for each one.

```python
BUDGET = {"Groceries": 200, "Transport": 50, "Eating out": 80, "Bills": 120, "Fun": 80}

def over_budget(totals: dict[str, float], budget: dict[str, float]):
    for cat, spent in totals.items():
        if spent > budget.get(cat, 0):
            yield f"{cat}: spent ${spent:.2f}, budget ${budget[cat]:.2f}, over by ${spent - budget[cat]:.2f}"

warnings = list(over_budget(totals, BUDGET))
print("\n".join(warnings) if warnings else "Everything is within budget!")
```

Before we write the report, we **test** our functions with a tiny, hand-made list where we know the answer.

```python
sample = [Expense("2026-01-01", "Fun", "a", 10.0), Expense("2026-01-02", "Fun", "b", 5.5),
          Expense("2026-01-03", "Bills", "c", 20.0)]
assert totals_by_category(sample) == {"Bills": 20.0, "Fun": 15.5}
assert list(over_budget({"Fun": 15.5}, {"Fun": 10})) == ["Fun: spent $15.50, budget $10.00, over by $5.50"]
print("Tracker tests passed!")
```

Finally, we write a markdown **report** file, using a `with` block and f-strings, then read it back.

```python
def write_report(path: str, items: list[Expense]) -> None:
    totals = totals_by_category(items)
    with open(path, "w") as f:
        f.write("# January spending report\n\n")
        f.write(f"Expenses: {len(items)}  |  Total: ${sum(totals.values()):.2f}\n\n")
        f.write("## By category\n")
        for cat, amt in totals.items():
            f.write(f"- {cat}: ${amt:.2f}\n")
        f.write("\n## Over budget\n")
        for w in over_budget(totals, BUDGET):
            f.write(f"- {w}\n")

write_report("report.md", expenses)
print(open("report.md").read())
```
